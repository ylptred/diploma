```python
def transcribe_and_score(audio_path, phrases_dict):
    """Компонент 1: Трансквилирование + поиск фраз"""
    try:
        # Загружаем первые 30 секунд
        y, sr = librosa.load(audio_path, sr=16000, duration=30)
        
        # Транскрибируем
        result = model_whisper.transcribe(audio_path, language="ru", fp16=False)
        transcript = result['text'].lower()
        
        # Ищем мошеннические фразы
        keyword_score = 0
        found = []
        
        for phrase, weight in phrases_dict.items():
            # Точное совпадение
            if phrase in transcript:
                keyword_score += weight
                found.append(phrase)
            # Нечёткое совпадение
            elif fuzz.token_set_ratio(phrase, transcript) > 85:
                keyword_score += weight * 0.7
        
        # Нормализуем [0, 1]
        keyword_score = min(keyword_score / len(phrases_dict), 1.0)
        
        return {
            'transcript': transcript[:200],
            'keyword_score': keyword_score,
            'found_phrases': found[:5],
            'risk_level': 'КРИТИЧЕСКИЙ' if keyword_score > 0.7 else 
                         'ВЫСОКИЙ' if keyword_score > 0.5 else
                         'СРЕДНИЙ' if keyword_score > 0.3 else 'НИЗКИЙ'
        }
    except Exception as e:
        print(f"   ❌ Ошибка: {e}")
        return None

# БЫСТРЫЙ ТЕСТ
if len(synthetic_paths) > 0:
    print("\n🧪 ТЕСТ КОМПОНЕНТА 1\n")
    result = transcribe_and_score(synthetic_paths[0], PHRASES)  # Берём первый файл
    if result:
        print(f" 📝 Транскрипция: {result['transcript'][:100]}...")
        print(f" 🎯 Keyword Score: {result['keyword_score']:.3f}")
        print(f" ⚠️ Уровень риска: {result['risk_level']}")
else:
    print("❌ Нет синтетических аудиофайлов для тестирования")

```

```python
from tqdm import tqdm

def extract_features_fast(audio_path):
    """Извлечение 7 просодических признаков"""
    try:
        snd = parselmouth.Sound(audio_path)
        pitch = snd.to_pitch(time_step=0.02)
        
        f0_values = pitch.selected_array['frequency']
        f0_values = f0_values[f0_values > 0]
        
        if len(f0_values) < 10:
            return None
        
        features = {
            'f0_mean': np.mean(f0_values),
            'f0_std': np.std(f0_values),
            'f0_range': np.max(f0_values) - np.min(f0_values),
            'f0_delta': np.mean(np.abs(np.diff(f0_values))),
            'num_voiced': len(f0_values),
            'jitter': np.mean(np.abs(np.diff(f0_values))) / np.mean(f0_values) if len(f0_values) > 1 else 0,
        }
        
        y, sr = librosa.load(audio_path, sr=16000)
        features['intensity_std'] = np.std(librosa.feature.melspectrogram(y=y, sr=sr))
        
        return features
    except:
        return None

print("🤖 ОБУЧЕНИЕ RANDOM FOREST (это займёт ~1.5 часа)\n")
print(f"   Синтетических файлов: {len(synthetic_paths)}")
print(f"   Естественных файлов: {len(natural_paths)}\n")

X = []
y = []

# СИНТЕТИЧЕСКИЕ (класс 1)
print("📂 Обработка синтетических файлов...")
successful = 0
failed = 0

pbar = tqdm(synthetic_paths, desc="🔴 Синтетика", unit="файл", colour="red")
for path in pbar:
    features = extract_features_fast(path)
    if features:
        X.append([features['f0_mean'], features['f0_std'], features['f0_range'], 
                 features['f0_delta'], features['jitter'], features['num_voiced'],
                 features['intensity_std']])
        y.append(1)
        successful += 1
    else:
        failed += 1
    pbar.set_postfix({"успешно": successful, "ошибки": failed})

# ЕСТЕСТВЕННЫЕ (класс 0)
print("\n📂 Обработка естественных файлов...")
successful = 0
failed = 0

pbar = tqdm(natural_paths, desc="🟢 Реальная речь", unit="файл", colour="green")
for path in pbar:
    features = extract_features_fast(path)
    if features:
        X.append([features['f0_mean'], features['f0_std'], features['f0_range'], 
                 features['f0_delta'], features['jitter'], features['num_voiced'],
                 features['intensity_std']])
        y.append(0)
        successful += 1
    else:
        failed += 1
    pbar.set_postfix({"успешно": successful, "ошибки": failed})


print(f"\n✅ Собрано {len(X)} образцов")

# ПОДГОТОВКА ДАННЫХ
X = np.array(X, dtype=np.float32)
y = np.array(y)

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# ОБУЧЕНИЕ (ДОЛГОЕ)
print("\n⏳ ОБУЧЕНИЕ RANDOM FOREST (это займёт 30-60 минут)...\n")

rf = RandomForestClassifier(
    n_estimators=80,           # Было бы 100-150, но 80 достаточно
    max_depth=12,              # Оптимально для 7 признаков
    min_samples_split=20,      # Увеличено для защиты от переобучения
    min_samples_leaf=10,       # Добавлено - минимум 10 образцов в листе
    max_features='sqrt',       # ~2.6 признаков на дерево (из 7)
    max_samples=0.8,           # Bootstrap 80% данных (ускорение)
    class_weight='balanced',
    n_jobs=-1,                 # Все доступные CPU-ядра
    verbose=2,                 # Подробный вывод
    random_state=42
)

rf.fit(X_scaled, y)

# КРОСС-ВАЛИДАЦИЯ
print("\n⏳ Кросс-валидация...")
cv_scores = cross_val_score(rf, X_scaled, y, cv=3, scoring='f1_weighted')

print(f"\n✅ ОБУЧЕНИЕ ЗАВЕРШЕНО!")
print(f"   CV F1-score: {cv_scores.mean():.3f} ± {cv_scores.std():.3f}")
print(f"   Train F1-score: {rf.score(X_scaled, y):.3f}")

# СОХРАНЯЕМ МОДЕЛЬ
prosodic_model = {
    'rf': rf,
    'scaler': scaler,
    'feature_names': ['f0_mean', 'f0_std', 'f0_range', 'f0_delta', 'jitter', 'num_voiced', 'intensity_std']
}

with open(f'{KAGGLE_BASE}/prosodic_model.pkl', 'wb') as f:
    pickle.dump(prosodic_model, f)

print(f"\n💾 Модель сохранена: {KAGGLE_BASE}/prosodic_model.pkl")
```

```python
def predict_synthetic(audio_path, model_data):
    """Предсказание вероятности синтетической речи"""
    try:
        features = extract_features_fast(audio_path)
        if features is None:
            return None
        
        X_test = np.array([[
            features['f0_mean'],
            features['f0_std'],
            features['f0_range'],
            features['f0_delta'],
            features['jitter'],
            features['num_voiced'],
            features['intensity_std']
        ]])
        
        X_scaled = model_data['scaler'].transform(X_test)
        proba = model_data['rf'].predict_proba(X_scaled)
        
        return {
            'prob_synthetic': float(proba),
            'prob_natural': float(proba),
            'confidence': float(max(proba))
        }
    except:
        return None

print("✅ Функция predict_synthetic() готова")
```

```python
def analyze_audio_complete(audio_path):
    """ПОЛНЫЙ АНАЛИЗ: Компонент 1 + Компонент 2"""
    
    print(f"\n🔍 АНАЛИЗ: {os.path.basename(audio_path)}")
    print("=" * 70)
    
    # КОМПОНЕНТ 1
    print("\n📝 КОМПОНЕНТ 1: Keyword Spotting")
    c1 = transcribe_and_score(audio_path, PHRASES)
    
    if c1:
        print(f"   Keyword Score: {c1['keyword_score']:.3f}")
        print(f"   Найдено фраз: {len(c1['found_phrases'])}")
        print(f"   Риск: {c1['risk_level']}")
    else:
        c1 = {'keyword_score': 0.5, 'risk_level': 'НЕИЗВЕСТЕН'}
    
    # КОМПОНЕНТ 2
    print("\n🎤 КОМПОНЕНТ 2: Просодический анализ")
    c2 = predict_synthetic(audio_path, prosodic_model)
    
    if c2:
        print(f"   P(синтетическая): {c2['prob_synthetic']:.3f}")
        print(f"   Уверенность: {c2['confidence']:.3f}")
    else:
        c2 = {'prob_synthetic': 0.5, 'confidence': 0.0}
    
    # СИНТЕЗ
    final_score = (c1['keyword_score'] + c2['prob_synthetic']) / 2
    final_level = 'КРИТИЧЕСКИЙ' if final_score > 0.75 else \
                 'ВЫСОКИЙ' if final_score > 0.55 else \
                 'СРЕДНИЙ' if final_score > 0.35 else 'НИЗКИЙ'
    
    emoji = {'КРИТИЧЕСКИЙ': '🔴', 'ВЫСОКИЙ': '🟠', 'СРЕДНИЙ': '🟡', 'НИЗКИЙ': '🟢'}
    
    print(f"\n🎯 ИТОГОВАЯ ОЦЕНКА")
    print(f"   {emoji.get(final_level, '❓')} {final_level}")
    print(f"   Риск-score: {final_score:.3f}")
    print("=" * 70)
    
    return {'c1': c1, 'c2': c2, 'final_score': final_score, 'final_level': final_level}

print("✅ Функция analyze_audio_complete() готова")
```

```python
print("🧪 ТЕСТИРОВАНИЕ МОДУЛЯ 2\n")

# Выбираем тестовые файлы
test_files = synthetic_paths[:2] + natural_paths[:1]

if len(test_files) > 0:
    for test_file in test_files:
        result = analyze_audio_complete(test_file)
else:
    print("⚠️  Нет файлов для тестирования")

print("\n✅ Тестирование завершено!")
```
